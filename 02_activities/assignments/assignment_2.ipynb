#### Starter Code for Question 2

```python
# Definition for a binary tree node.
class TreeNode(object):
  def __init__(self, val = 0, left = None, right = None):
    self.val = val
    self.left = left
    self.right = right

def bt_path(root: TreeNode) -> list[list[int]]:  
  
  def helper_bt_path(node, current_path, all_paths):
    if not node:
      return
    
    current_path.append(node.val)

    if not node.left and not node.right:
      all_paths.append(list(current_path))
    else:
      helper_bt_path(node.left, current_path, all_paths)
      helper_bt_path(node.right, current_path, all_paths)
    
    current_path.pop()

  result = []
  helper_bt_path(root,[],result)
  return result

```


-   Copy the solution your partner wrote. 


```python
# Your answer here
def move_zeros_to_end(nums: list[int]) -> list[int]:
    #start at index 0 on list, keeps track of where next number should be placed.
    write_pos = 0
    # First pass: Move all non zero elements to the front of the list, in range of the number of elements. 
    # length of list gives number of elements
    for read in range(len(nums)):
        #l if element is a non-zero, place in next write_pos
        if nums[read] != 0:
            nums[write_pos] = nums[read]
            #move write_pos forward one index
            write_pos += 1

    #Second pass: fill the remaining positions with zeros
    while write_pos < len(nums):
        nums[write_pos]=0
        write_pos+=1

    #return the modified list 
    return nums
```


-   Explain why their solution works in your own words.


```python
# Your answer here
# Their solution works by using a for loop to first modify the list with only the non-zero elements,
# and the for loop ensures this list is created in the same order as the original list.
# Once the list is passed through and all its non-zero elements are moved to be first in the list,
# the list is filled in with zeros for the rest of the length of the list.
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Your answer here
# The time complexity is O(n) because the original list length is only iterated through twice.
# The space complexity is O(1) because the original list is being modified - so no extra space is required.
```


-   Critique your partner's solution, including explanation, and if there is anything that should be adjusted.
